```python
project_id = "project_123"
vectorstore = BQVectorStore()
query = "Hvem sa at "
k = 5
results = vectorstore.query(query=query, collection_id="attachments", k=k, filter = {"project_id": project_id})
if not results:
    result =  f"No relevant information found in the vectorstore for project {project_id}."
retrieved_content = "\n".join([f"- {doc.page_content}" for doc in results])
result = f"Retrieved information from vectorstore for project {project_id}:\n{retrieved_content}"
```

```python
vectorstore = BQVectorStore()
query = '''Jeg har nylig kjøpt en bolig og nå dukker det opp masse feil ved boligen. Er kjøpt etter avhendingsloven, brukt bolig.
Hva er min reklamasjonsfrist for skjulte feil? og hva kan jeg kreve og ikke kreve?'''
k = 5
results = vectorstore.query(query=query, collection_id="laws", k=k)
if not results:
    res =  "No relevant laws found in the vectorstore."
res = "Retrieved relevant laws from vectorstore:\n"
for doc in results:
    res += f'Legal area: {doc.metadata.get("legal_area", "Unknown")} | Title {doc.metadata.get("title", "Unknown")} | Paragraph {doc.metadata.get("paragraph_number", "Unknown")}\n'
    res += f"- {doc.page_content}\n"
retrieved_content = "\n".join([f"- {doc.page_content}" for doc in results])
res =  f"\n{retrieved_content}"
```
