```python
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_google_community import BigQueryVectorStore
import os
from dotenv import load_dotenv
from langchain_chroma import Chroma
load_dotenv()
```

```python
dataset: str = "vector_store"
REGION: str = "europe-north2"
model_name: str = "text-embedding-004"
PROJECT_ID = os.getenv("GOOGLE_CLOUD_PROJECT")
DATASET = "vector_store"
TABLE = "attachments"
embedding = GoogleGenerativeAIEmbeddings(model=model_name)
splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap=200)
```

```python
vs_chroma = Chroma(collection_name="my_collection", 
                   embedding_function=embedding)
```

```python
vs_bq = BigQueryVectorStore(
                project_id=PROJECT_ID,
                dataset_name=DATASET,
                table_name=TABLE,
                location=REGION,
                embedding=embedding,
            )
```

```python
file = '/Users/sigvardbratlie/Documents/Projects/master-thesis/data/THRD-2021-163881/01_fabricated/2025-07-16_132_Domisivilsak-fagdommer(e)_242348.txt'
with open(file, 'r') as f:
    text = f.read()

docs = splitter.create_documents([text])
```

```python
vs_chroma.add_documents(docs)
```

```python
vs_bq.add_documents(docs)
```
