```python
import re
from pathlib import Path
from datetime import datetime
import os
start_date = datetime(2025,10,31)
end_date   = None

files = []
pattern = re.compile(r'\d{4}-\d{2}-\d{2}')

for file in Path("./02_org_pdf").glob("*"):
    match = pattern.search(file.name)
    if not match:
        continue
    
    try:
        file_date_str = match.group(0)
        file_date = datetime.fromisoformat(file_date_str)
    except ValueError:
        continue  # ugyldig dato → hopp over

    include = True  # default: ta med

    if start_date is not None and file_date < start_date:
        include = False
    
    if end_date is not None and file_date > end_date:
        include = False

    if include:
        files.append(str(file))

files.sort()  # sorterer etter dato, siden datoen er i YYYY-MM-DD-format
```
