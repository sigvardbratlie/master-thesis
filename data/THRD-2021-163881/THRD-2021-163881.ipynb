## Token Count

```python
tot = 0
for file in Path("data/THRD-2021-163881/deanonymized/").glob("*.txt"):
    with open(file, "r") as f:
        text = f.read()
    
    encoding = tiktoken.get_encoding("gpt2")
    tokens = encoding.encode(text)
    print(f"{file.name}: {len(tokens)} tokens")
    tot += len(tokens)
print(f"Total tokens: {tot}")
```
